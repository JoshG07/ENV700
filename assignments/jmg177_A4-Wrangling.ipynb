### 1. Set up your Python Session

#### ✅ 1.1 Import Packages

````python
# 1.1 Import Packages
from pathlib import Path
import pandas as pd
````

* Display the the values of these three variables in your notebook.

````python
# 1.2 Set project folders using relative paths
root_fldr = Path.cwd().parent
raw_fldr = root_fldr / "data" / "raw"
processed_fldr = root_fldr / "data" / "processed"
````

#### ✅ 1.3 Set a path to the raw data files
* Create a Path variable pointing to the four EPA AQS files found in the raw data folder: 
    - `EPAair_O3_NC2024_raw.csv`
    - `EPAair_O3_NC2025_raw.csv`
    - `EPAair_PM25_NC2024_raw.csv`

````python
# 1.3 Create file path variables
ozone_2024_file = raw_fldr / "EPAair_O3_NC2024_raw.csv"
ozone_2025_file = raw_fldr / "EPAair_O3_NC2025_raw.csv"
pm25_2024_file = raw_fldr / "EPAair_PM25_NC2024_raw.csv"
pm25_2025_file = raw_fldr / "EPAair_PM25_NC2025_raw.csv"
````

(*No need to set data types, parse dates, or perform any other specialize import functions here; we will wrangle value later.*)

````python
# 1.4 Read data into dataframes
ozone_2024 = pd.read_csv(ozone_2024_file)
ozone_2025 = pd.read_csv(ozone_2025_file)
pm25_2024 = pd.read_csv(pm25_2024_file)
pm25_2025 = pd.read_csv(pm25_2025_file)
````

````python
# 1.5 Report the dimensions of each dataset
print("Ozone 2024:", ozone_2024.shape,
      "\nOzone 2025:", ozone_2025.shape,
      "\nPM 2.5:", pm25_2024.shape,
      "\nPM 2.5:", pm25_2025.shape)
````

### 2. Wrangle individual datasets

#### ✅ 2.1 Update data types
* Convert the `Date` column in each dataframe so that values are datetime objects. 
* List the date range for the two Ozone dataframes.

````python
# Convert the Date column to DateTime objects
ozone_2024["Date"]

ozone_2024["Date"] = pd.to_datetime(ozone_2024["Date"], format = "%m/%d/%Y")
ozone_2025["Date"] = pd.to_datetime(ozone_2025["Date"], format = "%m/%d/%Y")
pm25_2024["Date"] = pd.to_datetime(pm25_2024["Date"], format = "%m/%d/%Y")
pm25_2025["Date"] = pd.to_datetime(pm25_2025["Date"], format = "%m/%d/%Y")

# Print date ranges
print("Ozone 2024:", ozone_2024["Date"].min(), "-", ozone_2024["Date"].max())
print("Ozone 2025:", ozone_2025["Date"].min(), "-", ozone_2025["Date"].max())
````

#### ✅ 2.2 Eliminate unnecessary columns
* Subset each dataframe to include only the following columns:  
```'Date', 'Daily AQI Value', 'Local Site Name', 'AQS Parameter Description', 'County', 'Site Latitude', 'Site Longitude'```

````python
# 2.2 Subset columns
columns_subset = ['Date', 'Daily AQI Value', 'Local Site Name', 'AQS Parameter Description', 'County', 'Site Latitude', 'Site Longitude']

ozone_2024 = ozone_2024.loc[:, columns_subset]
ozone_2025 = ozone_2025.loc[:, columns_subset]
pm25_2024 = pm25_2024.loc[:, columns_subset]
pm25_2025 = pm25_2025.loc[:, columns_subset]

pm25_2025.head()
````

#### ✅ 2.3 Update values
* For the PM2.5 datasets, fill all cells in the `AQS Parameter Description` with the string "PM2.5".  
(*All cell values in this column will be identical.*)

````python
# 2.3 Update values
pm25_2024["AQS Parameter Description"] = "PM2.5"
pm25_2025["AQS Parameter Description"] = "PM2.5"

pm25_2024.head()
````

### 3. Combine datasets

#### ✅ 3.1 Confirm dataframes are set to be merged
* Ensure that column names in each dataframe are indentical

````python
# 3.1 Ensure column are same across all dataframes
print(ozone_2024.columns,
ozone_2025.columns,
pm25_2024.columns,
pm25_2025.columns)
````

#### ✅ 3.2 Combine the datasets 
* Combine the datasets using Pandas' `concat()` function
* Report the dimensions of the merged dataframe. (This should be `38,921 x 7`)

````python
# 3.2 Concatentate & report shape
full_AQI = pd.concat(
    [ozone_2024, ozone_2025, pm25_2024, pm25_2025],
    ignore_index=True
)

full_AQI.shape
````

#### ✅ 3.3 Examine the contents of the merged dataframe
* Display a sample of 10 randomly selected rows, using the random state = 42

````python
# 3.3 View a sample of data
full_AQI.sample(10, random_state=42)
````

### 4. Wrangle the combined dataset

#### ✅ 4.1 Subset rows 
* Generate a list of local site names common to all 4 dataframes (code provided, but replace the dataframe names to match your own).
* Create a subset of the full dataset, selecting those that match these local site names.
* Report the dimensions of the resulting dataframe. (This should be `18,778 x 7`)

````python
# Generate a list of local site names common to all 4 dataframes
## Create set objects from the values in the Local Site Name columns
sites_oz24 = set(ozone_2024['Local Site Name']) 
sites_oz25 = set(ozone_2025['Local Site Name'])
sites_pm24 = set(pm25_2024['Local Site Name'])
sites_pm25 = set(pm25_2025['Local Site Name'])
## Compute the intersection 
common_sites = sites_oz24 & sites_oz25 & sites_pm24 & sites_pm25
common_sites
````

````python
# 4.1 Subset for records with sites in common
site_subset_AQI = full_AQI.loc[full_AQI["Local Site Name"].isin(common_sites)]

site_subset_AQI.shape
````

#### ✅ 4.2 Group and summarize data 
We want a dataset that lists one record per measured pollutant per location per day. To ensure we have that, we'll group data by date, local site name, AQS parameter, and county, computing the mean daily AQI value as well as mean latitude and longitude value. 
* Group the data on the grouping variables mentioned above. 
* For each group, compute mean daily AQI value, mean latitude, and mean longitude. 
* Reset the index of the summarized dataset to restore the product to a flat dataframe.
* Report the dimensions of the dataset. (This should be `17,017 x 7`.)

````python
# 4.2 Group and summarize data, then report its dimensions
summarized_AQI = (
    site_subset_AQI
    .groupby(["Date", "Local Site Name", "AQS Parameter Description", "County"])
    .agg(

        mean_AQI = ("Daily AQI Value", "mean"),
        mean_latitude = ("Site Latitude", "mean"),
        mean_longitude = ("Site Longitude", "mean")
    )
    .reset_index()
)

summarized_AQI.shape
````

#### ✅ 4.3 Add calculated columns
* Add `Year` and `Month` columns to your dataframe, deriving values from the `Date` field. 

````python
# 4.3 Calculate year and month columns
summarized_AQI["year"] = summarized_AQI["Date"].dt.year
summarized_AQI["month"] = summarized_AQI["Date"].dt.month

summarized_AQI.head()
````

#### ✅ 4.4 Spread AQI values into separate columns: OZone & PM2.5
* Reshape our table so that Daily AQI values for Ozone and PM2.5 data appear in separate columns.
* Reset the index of the reshaped table to restore it as a flat dataframe.
* Display the first 5 records in the resulting dataframe 

````python
# 4.4 Spread data so Ozone, PM2.5 daily AQI valiues occupy separate columns
summarized_AQI_wide = (
    summarized_AQI
    .pivot_table(
        index = ["Date", "Local Site Name", "County", "month", "year"],
        columns = "AQS Parameter Description",
        values = "mean_AQI",
        aggfunc = "mean"
    )
    .reset_index()
)

summarized_AQI_wide.head()
````

````python
summarized_AQI_wide["Ozone"].notna().value_counts()
````

## 6. Summarize and explore the data

#### ✅ 6.1 Generate a month summary table
* Group the above table to compute mean Ozone and PM2.5 values for each unique site, month, year combination. Be sure to reset the index of the product. 
* Display the first 5 records of this table.

````python
# 6.1 Generate a summary table and show its first 5 records
summary_table = (
    summarized_AQI_wide
    .groupby(["Local Site Name", "month", "year"])
    .agg(
        mean_ozone = ("Ozone", "mean"),
        mean_pm25 = ("PM2.5", "mean")
    )
    .reset_index()
)

summary_table.head()
````
